```python
X = df.drop('price', axis = 1)
y = df['price']
```

```python
X_onehot_encoded = pd.get_dummies(X, columns = ['model','transmission','fuelType'])
```

```python
X_onehot_encoded = X_onehot_encoded.astype(int)
```

```python
X_onehot_encoded
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encode = LabelEncoder()
```

```python
Xlabel = X
```

```python
column = ['model','transmission','fuelType']
```

```python
for i in column:
  Xlabel[i] = encode.fit_transform(Xlabel[i])
```

```python
Xlabel
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_columns = ['year','mileage','tax','mpg','engineSize']
```

```python
scalar = StandardScaler()
```

```python
X_onehot_encoded[numerical_columns] = scalar.fit_transform(X_onehot_encoded[numerical_columns])
```

```python
X_onehot_encoded
```

```python
Xlabel.columns
```

```python
Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] =scalar.fit_transform(Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])
```

```python
Xlabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error , mean_squared_error
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_onehot_encoded, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test, y_pred)
r2
```
