```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
model_knn = KNeighborsClassifier(n_neighbors=25)
```

```python
model_knn.fit(X_train, y_train)
```

```python
model_knn.score(X_test, y_test)
```

```python
model_svm = SVC(C = 30,kernel='rbf',gamma='auto')
model_svm.fit(X_train, y_train)
```

```python
model_svm.score(X_test, y_test)
```

# now using grid serach CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_svm),{
    'C' : [1,10,20,30],
    'kernel' : ['rbf','linear']
},cv=5,return_train_score=False)
```

```python
classifier.fit(X,y)
```

```python
print(classifier.cv_results_)
```

```python
results = pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

```python
classifier_1 = GridSearchCV(model_knn,{
    'n_neighbors' : [3,5,7,9,11],
    'weights' : ['uniform','distance'],
    'metric' : ['euclidean','manhattan']
},cv=5,return_train_score=False)
```

```python
classifier_1.fit(X,y)
```

```python
results_1 = pd.DataFrame(classifier_1.cv_results_)
```

```python
results_1
```

```python
results_1[['param_n_neighbors','param_weights','param_metric','mean_test_score']]
```
