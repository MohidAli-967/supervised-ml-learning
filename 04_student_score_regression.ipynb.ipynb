```python
X = df.drop('math score', axis=1)
y = df['math score']
```

```python
df.columns
```

```python
X_onehot_encoded = pd.get_dummies(X, columns = ['race/ethnicity','gender','parental level of education','lunch','test preparation course','college_name'])
```

```python
X_onehot_encoded = X_onehot_encoded.astype(int)
```

```python
X_onehot_encoded
```

```python
from sklearn.preprocessing import LabelEncoder
encode = LabelEncoder()
```

```python
Xlabel = X
```

```python
column = ['race/ethnicity','gender','parental level of education','lunch','test preparation course','college_name']
for col in column:
  Xlabel[col] = encode.fit_transform(Xlabel[col])
```

```python
Xlabel
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_columns = ['writing score','reading score']
```

```python
scalar = StandardScaler()
```

```python
X_onehot_encoded[numerical_columns] = scalar.fit_transform(X_onehot_encoded[numerical_columns])
```

```python
X_onehot_encoded
```

```python
Xlabel[['race/ethnicity','gender','parental level of education','lunch','test preparation course','reading score'
,'writing score','college_name'] ] = scalar.fit_transform(Xlabel[['race/ethnicity','gender','parental level of education','lunch','test preparation course',
        'reading score','writing score','college_name']])
```

```python
Xlabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_onehot_encoded, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test, y_pred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1-(1-r2)*(n-1)/(n-p-1)
adjusted_r2
```
